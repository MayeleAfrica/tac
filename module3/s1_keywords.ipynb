## Imports

```python
import os
import yake
```

## Extraire les mots clés d'un document avec Yake

https://github.com/LIAAD/yake

```python
# Instantier l'extracteur de mots clés
kw_extractor = yake.KeywordExtractor(lan="fr", top=50)
kw_extractor
```

```python
# Lister les Fichiers
# data_path = "../data/txt/"
# files = [f for f in os.listdir(data_path) if f.endswith('.txt')]
# Lister les fichiers de 1913
from pathlib import Path

# Le notebook peut être lancé depuis la racine du dépôt ou depuis module3.
if Path("data/txt").is_dir():
    data_path = Path("data/txt")
elif Path("../data/txt").is_dir():
    data_path = Path("../data/txt")
else:
    raise FileNotFoundError("Le dossier data/txt est introuvable.")

files = sorted(
    file.name
    for file in data_path.glob("*.txt")
    if "1913" in file.name
)

```

```python
# Imprimer le nombre de fichiers identifiés
print(f"Fichiers trouvés pour 1913 : {len(files)}")
```

```python
# Les dix premiers fichiers

files[:10]
```

```python
from collections import Counter

keyword_counts = Counter()

for file_name in files:
    with open(data_path / file_name, "r", encoding="utf-8") as file:
        text = file.read()

    keywords = kw_extractor.extract_keywords(text)

    # On garde les expressions de deux mots, une seule fois par fichier.
    bigrams = {
        keyword
        for keyword, score in keywords
        if len(keyword.split()) == 2
    }
    keyword_counts.update(bigrams)

# Afficher les 30 expressions trouvées dans le plus grand nombre de fichiers
for keyword, count in keyword_counts.most_common(30):
    print(f"{keyword} : présent dans {count} fichiers")
```

```python
# Choisir un fichier
this_file = files[0]
this_file
```

```python
# Récupérer le texte du fichier
text = open(os.path.join(data_path, this_file), 'r', encoding='utf-8').read()
text[:500]
```

```python
# Extraire les mots clés de ce texte
keywords = kw_extractor.extract_keywords(text)
keywords
```

```python
# Ne garder que les bigrammes
kept = []
for kw, score in keywords:
    words = kw.split()
    if len(words) == 2:
        kept.append(kw)
kept
```

## Faire la même opération sur tous les documents

```python
for f in sorted(files)[:10]:
    text = open(os.path.join(data_path, f), 'r', encoding="utf-8").read()
    keywords = kw_extractor.extract_keywords(text)
    kept = []
    for kw, score in keywords:
        words = kw.split()
        if len(words) == 2:
            kept.append(kw)
    print(f"{f} mentions these keywords: {', '.join(kept)}...")
```
