## Appliquer la reconnaissance d'entités nommées sur notre corpus

```python
# Charger le texte
# n=1000000
# text = open("../data/all.txt", encoding='utf-8').read()[:n]
from pathlib import Path

# Trouver le dossier du corpus
if Path("data/txt").is_dir():
    txt_path = Path("data/txt")
elif Path("../data/txt").is_dir():
    txt_path = Path("../data/txt")
else:
    raise FileNotFoundError("Le dossier data/txt est introuvable.")

# Sélectionner les fichiers de 1913
year = 1913
files = sorted(txt_path.glob(f"*{year}*.txt"))
print(f"Fichiers trouvés pour {year} : {len(files)}")

# Préparer les compteurs
entities = {
    "personnes": defaultdict(int),
    "organisations": defaultdict(int),
    "lieux": defaultdict(int),
}

# Analyser chaque fichier
for file_path in files:
    text = file_path.read_text(encoding="utf-8")
    doc = nlp(text)

    for ent in doc.ents:
        name = ent.text.strip()

        if len(name) <= 3:
            continue
        if ent.label_ == "PER":
            entities["personnes"][name] += 1
        elif ent.label_ == "ORG":
            entities["organisations"][name] += 1
        elif ent.label_ in ("LOC", "GPE"):
            entities["lieux"][name] += 1

# Afficher les 20 entités les plus fréquentes par catégorie
for category, counts in entities.items():
    print(f"\n{category.capitalize()} :")
    for name, count in sorted(counts.items(), key=lambda item: item[1], reverse=True)[:20]:
        print(f"{name} : {count}")
```
