## Imports

```python
import json
import requests
```

## Utiliser [Nominatim](https://nominatim.org/) pour connaître les coordonnées géographiques d'une adresse

Le document `data/camille/KB_JB230_1913-07-05_01-0001.pdf` mentionne Bruxelles. Nous interrogeons l'API Nominatim pour obtenir les coordonnées géographiques actuelles de la ville.

```python
place = "Bruxelles, Belgique"

"""Retrieve coordinates from OpenStreetMap using Nominatim."""
headers = {"User-Agent": "MayeleAfrica-TAC-TP1/1.0"}
url = "https://nominatim.openstreetmap.org/search"
data = {
    "q": place,
    "format": "jsonv2",
    "limit": 1,
    "accept-language": "fr",
}

resp = requests.get(url, params=data, headers=headers, timeout=30)
resp.raise_for_status()

try:
    json_list = json.loads(resp.text)
except json.JSONDecodeError:
    print("La réponse reçue n'est pas au format JSON.")
    print(resp.text)
else:
    if json_list:
        for item in json_list:
            display_name = item["display_name"]
            lat = item["lat"]
            lon = item["lon"]
            print(f"{display_name} ({lat}, {lon})")
    else:
        print(f"Aucun résultat trouvé pour : {place}")
```
